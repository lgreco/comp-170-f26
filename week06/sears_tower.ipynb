# The Sears Tower (aka Willis Tower)

**Basic dimensions**

* Height: 1425 feet
* Width: 225 feet
* $\text{aspect ratio} = \dfrac{\text{height}}{\text{width}} = 6.3$

![](./sears_drawing.png)

```python
# Sears tower facts

HEIGHT = 1_425
WIDTH = 225
ASPECT_RATIO = HEIGHT / WIDTH
```

```python
def draw_sears_tower(height: int, aspect_ratio: float, building_block: str):
    """Draws the Sears Tower with the specified height and building block.

    Inputs:
    -------
    height: The height of the Sears Tower.
    aspect_ratio: The aspect ratio of the Sears Tower.
    building_block: The character to use for the building block.
    """ 
    # Protect agaist bad aspect ratio. The aspect ratio should be a positive 
    # number and greater than zero. If it is not, we will default to 1.0.  
    if aspect_ratio <= 0:
        aspect_ratio = 1.0
    # Given the height and aspect ratio, calculate the width
    width = int(height / aspect_ratio)
    # Assumptions to take into account:
    # The Sears Tower is composed of multiple tiers. Here we simplify it to three
    # tiers: bottom, middle, and top. Each tier has a progressively smaller width and
    # height.
    # --- Bottom tier: full width, about 60% of the total height
    # --- Middle tier: 80% of the bottom tier's width, 25% of the total height
    # --- Top tier: 80% of the bottom tier's width, 15% of the total height
    # 
    # Commit these assumptions to constants for clarity and maintainability
    MIDDLE_TIER_WIDTH_SCALE = 0.8
    TOP_TIER_WIDTH_SCALE = 0.8 * MIDDLE_TIER_WIDTH_SCALE
    BOTTOM_TIER_HEIGHT_SCALE = 0.6
    MIDDLE_TIER_HEIGHT_SCALE = 0.25
    TOP_TIER_HEIGHT_SCALE = 0.15
    MIN_ACCEPTABLE_WIDTH = 9
    # Compute tier-specific widths and heights
    bottom_tier_height = int(height * BOTTOM_TIER_HEIGHT_SCALE)
    bottom_tier_width = width   
    middle_tier_height = int(height * MIDDLE_TIER_HEIGHT_SCALE)
    middle_tier_width = int(width * MIDDLE_TIER_WIDTH_SCALE)
    top_tier_height = int(height * TOP_TIER_HEIGHT_SCALE)
    top_tier_width = int(width * TOP_TIER_WIDTH_SCALE)
    # print the top tier
    for floor in range(top_tier_height):
        print(building_block * top_tier_width)
    for floor in range(middle_tier_height):
        print(building_block * middle_tier_width)
    for floor in range(bottom_tier_height):
        print(building_block * bottom_tier_width)
```

```python
draw_sears_tower(30, 6.3, '#')
```
