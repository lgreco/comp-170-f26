# A Brief Note on Working with Lists

As we discussed in class, Python does not have *proper* arrays. Instead, we *pretend* that lists are arrays. For practical and instructional purposes, this is a reasonable simplification, and we will use the two terms interchangeably.

To see why arrays are useful, consider tracking the daily temperatures over a 7-day period:

````python
temperatures = [ 55, 61, 62, 63, 54, 65, 66 ] 
````

With the temperatures stored in a single list, we can iterate over it to compute their average.

### Iterating over a list with `for` and `range`

A list stores its elements in order, and each element has an *index*: the first element is at index `0`, the second at index `1`, and so on. A list with $n$ elements has valid indices $0$ through $n-1$.

To visit every element, we combine a `for` loop with `range`:

- `len(temperatures)` gives the number of elements, which is $n$.
- `range(n)` produces the integers $0, 1, \ldots, n-1$, exactly the valid indices.
- On each pass, the loop variable `i` takes the next index, and `temperatures[i]` is the element stored there.

Because the loop is driven by `len`, it adapts automatically to a list of any size. Note that `range(n)` stops *before* `n`; an index of `n` would be past the end of the list and raise an `IndexError`.

In the next cell, we use this pattern to add up the temperatures one element at a time.

````python
number_of_days = len(temperatures)
sum_of_temperatures = 0
for i in range(number_of_days):
    sum_of_temperatures = sum_of_temperatures + temperatures[i]
average_temperature = sum_of_temperatures / number_of_days
print("The average temperature is:", average_temperature)
````

As we discussed in class, this code works for an array of *any size*. Whether we track temperatures for 7 days or 700, the same code computes the average without modification.

Since the code is general, we can package it as a function of its own:

````python
def compute_average(values):
    number_of_values = len(values)
    sum_of_values = 0
    for i in range(number_of_values):
        sum_of_values = sum_of_values + values[i]
    average_value = sum_of_values / number_of_values
    return average_value
````

With this function, the computation reduces to a single call:

````python
average_temp = compute_average(temperatures)
print("The average temperature is:", average_temp)
````

If you prefer more polished output, an f-string can format the result to one decimal place:

````python
print(f"The average temperature is: {compute_average(temperatures):.1f}°F")
````

### Another way to iterate: looping over values

So far, we have used `range` to visit each *index* and then looked up the element at that index. Python also lets a `for` loop hand us the *values* directly:

```python
sum_of_temperatures = 0
for temperature in temperatures:
    sum_of_temperatures = sum_of_temperatures + temperature
```

Here, on each pass, `temperature` is the next element of the list itself. There is no index, no call to `len`, and no `temperatures[i]` lookup, so the loop is shorter and harder to get wrong.

An analogy may help. Imagine a list as a row of chairs at a long table, with one person in each chair:

| Chair 0 | Chair 1 | Chair 2 |
|:---:|:---:|:---:|
| Leo | Saruman | Sauron |

- Looping with `range` is like going **by seat position**: "Who is in chair 0? Say hello to that person. Now, who is in chair 1? Say hello to that person, etc." We count off the chairs, and for each chair we look to see who is sitting there. That's like writing
```python
for i in range(len(chairs)):
    name = chairs[i]
    print("Hello", name)
```
- Looping over values is like going **by identity**: "Leo, hello. Saruman, hello. Sauron, hello." We greet each person directly and never mention a chair number. That's like writing
```python
for name in chairs:
    print("Hello", name)
```

Either way, everyone at the table gets greeted, in the same order. The difference is whether we care about the *seat* or only about the *person* sitting in it.

Which style should you use? A good rule of thumb:

- If you only need the **values**, as when computing a sum or an average, looping over values is simpler and reads naturally.
- If you need the **position** of each element, as when comparing neighbors, tracking where a maximum occurs, or changing elements in place, loop over `range(len(...))` so that you have the index `i`.

Both loops visit the same elements in the same order, so for computing an average they produce identical results. For now, practice with both styles; knowing when each fits is part of writing clear code.
